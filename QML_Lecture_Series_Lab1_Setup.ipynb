## 2. Create and Draw Your First Quantum Circuit
A quantum circuit is a computational routine consisting of coherent quantum operations on qubits. Let's create an empty circuit with **1 qubit** and **1 classical bit**.

```python
from qiskit import QuantumCircuit

# Create a Quantum Circuit with 1 qubit and 1 classical bit
qc = QuantumCircuit(1, 1)

# Draw the empty circuit
qc.draw('mpl')
```

## 3. Compare a Classical Bit and a Qubit
A classical bit is always either `0` or `1`. A qubit starts in the ground state $|0\rangle$. Let's measure our qubit in its default state to see what happens.

```python
# Add a measurement operation
# We measure qubit 0 and store the result in classical bit 0
qc.measure(0, 0)

qc.draw('mpl')
```

## 4. Simulate a Single Qubit
To see the results of our measurement, we will run this circuit on a local simulator.

```python
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# Initialize the simulator
simulator = AerSimulator()

# Run the circuit 1000 times (shots)
job = simulator.run(qc, shots=1000)
result = job.result()

# Get the counts (how many times we measured 0 vs 1)
counts = result.get_counts(qc)
print("Measurement results:", counts)

plot_histogram(counts)
```

**Observation:** As expected, it behaves exactly like a classical bit initialized to 0! It measured `0` exactly 1000 times.

## 5. Run a Simple Hadamard Gate Experiment
Now let's see what makes a qubit special. We will apply a **Hadamard (H) gate**, which puts the qubit into a state of *superposition*.

```python
# Create a new circuit
qc_superposition = QuantumCircuit(1, 1)

# Apply Hadamard gate to qubit 0
qc_superposition.h(0)

# Measure the qubit
qc_superposition.measure(0, 0)

qc_superposition.draw('mpl')
```

```python
# Simulate the superposition circuit
job_sup = simulator.run(qc_superposition, shots=1000)
result_sup = job_sup.result()
counts_sup = result_sup.get_counts(qc_superposition)

print("Superposition results:", counts_sup)
plot_histogram(counts_sup)
```
